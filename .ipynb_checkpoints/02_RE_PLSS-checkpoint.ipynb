[Regular Expressions](https://docs.python.org/3/library/re.html) (regex) are a compact language for searching, matching, and manipulating text. Regex is widely used in Python, cleaning data and text processing, because a single expression can often replace many lines of string-handling code.

| Character | Meaning                         | Example           | Matches                 |
| --------- | ------------------------------- | ----------------- | ----------------------- |
| `.`       | Any character except newline    | `a.c`             | `abc`, `a7c`            |
| `^`       | Start of string/line            | `^Hello`          | `Hello world`           |
| `$`       | End of string/line              | `world$`          | `Hello world`           |
| `*`       | Zero or more repetitions        | `ab*c`            | `ac`, `abc`, `abbbc`    |
| `+`       | One or more repetitions         | `ab+c`            | `abc`, `abbbc`          |
| `?`       | Zero or one repetition          | `colou?r`         | `color`, `colour`       |
| `{n}`     | Exactly *n* repetitions         | `\d{4}`           | `2025`                  |
| `{n,m}`   | Between *n* and *m* repetitions | `\d{2,4}`         | `12`, `2025`            |
| `[]`      | Character class                 | `[aeiou]`         | Any vowel               |
| `[^]`     | Negated character class         | `[^0-9]`          | Any non-digit           |
| `\d`      | Digit                           | `\d+`             | `123`                   |
| `\D`      | Non-digit                       | `\D+`             | `abc`                   |
| `\w`      | Word character (`a-zA-Z0-9_`)   | `\w+`             | `hello_123`             |
| `\W`      | Non-word character              | `\W+`             | `@#$`                   |
| `\s`      | Whitespace                      | `\s+`             | spaces, tabs            |
| `\S`      | Non-whitespace                  | `\S+`             | `word`                  |
| `\b`      | Word boundary                   | `\bcat\b`         | `cat` but not `catalog` |
| `()`      | Capture group                   | `(\d{3})-(\d{4})` | Captures parts          |
| `\|`      | OR operator                     | `cat\|dog`         | `cat` or `dog`          |
| `\\`      | Escape special character        | `\.`              | Literal `.`             |


Combining `re` with the PLSS system allows us to make spatial representations of the text. We'll look at a simple example of geolocation conservation dockets from the Oklahoma Conservation Commission.

```python
import re
import geopandas as gpd
import numpy as np
import pandas as pd
import pytesseract
from pdf2image import convert_from_path
from PIL import Image
import glob

pd.set_option('display.max_columns', None)
```

[PyTesseract Installation](https://github.com/madmaze/pytesseract) instructions.

```python
sections = gpd.read_file("./data/Sections.geojson")
sections.head(2)
```

```python
def ocr_top_half_first_page(pdf_path, dpi=300):
    # Render first page as PIL image
    page = convert_from_path(
        pdf_path,
        dpi=dpi,
        first_page=1,
        last_page=1
    )[0]

    width, height = page.size

    # Crop top half
    top_half = page.crop((0, 0, width, height // 2))

    # OCR
    text = pytesseract.image_to_string(top_half)

    return text
```

```python
# make list of files
spath = './data/OCC_CD/*.pdf'
files = [x.replace('\\','/') for x in glob.glob(spath)]
print(len(files))

# Make a dataframe to keep this all organized
df = pd.DataFrame({'path':files})

# Open PDF, take top of 1st page, and OCR
for index, row in df.iterrows():
    txt = ocr_top_half_first_page(row['path'])
    df.at[index, 'txt'] = txt
    print('.', end='') #k.i.s.s. status indicator

df.head(2)
```

```python
for index, row in df.iterrows():
    txt = row['txt']
    # Clean up the new line characters
    txt = txt.replace('\n',' ')
    while '  ' in txt:
        txt = txt.replace('  ',' ')
    
    i0 = i1 = None
    
    # Find County
    pattern = ' [\w]+ COUNTY'
    match = re.search(pattern, txt)
    if match:
        county = match.group().replace('COUNTY','').strip()
        i0,i1 = match.span()
        df.at[index, 'county'] = county

        # Meridian
        if county in ['CIMARRON','TEXAS','BEAVER']:
            meridian = 'C'
        else:
            meridian = 'I'
        df.at[index, 'meridian'] = meridian
    
    # Find PLSS - handles "junk" in the ocr
    pattern = 'SECTION\s*(\d+).*?TOWNSHIP\s*(\d+).*?(NORTH|SOUTH).*?RANGE\s*(\d+).*?(EAST|WEST)'
    match = re.search(pattern, txt)
    if match:
        raw_plss = match.group()
        i0,i1 = match.span()
        df.at[index, 'plssraw'] = raw_plss

        pattern = '\d+'
        match = re.findall(pattern, raw_plss)
        if match:
            sec, t, r = match
            df.at[index, 'sect'] = sec
            df.at[index, 'township'] = t
            df.at[index, 'range'] = r

        # Directions
        if 'NORTH' in raw_plss:
            df.at[index,'t_d'] = 'N'
        else:
            df.at[index,'t_d'] = 'S'

        if 'EAST' in raw_plss:
            df.at[index,'r_d'] = 'E'
        else:
            df.at[index,'r_d'] = 'W'
            
    print('.', end='') #k.i.s.s. status indicator

df
```

```python
## Format text in columns

def lenstr(txt,n=2):
    '''
    Adds zeros to make string n len.
    '''
    txt = str(txt)
    while len(txt) < n:
        txt = '0' + txt
    return txt

for col in ['sect','township','range']:
    df[col] = df[col].apply(lenstr)
    
# Combine columns for merge column
# secttownran
df['secttownran'] = df['sect'] + ' ' + df['township'] + df['t_d'] + ' ' + df['range'] + df['r_d'] + df['meridian']

df
```

```python
#Merge gdf Section
gdf = df.merge(sections[['secttownran','geometry']], how='left')
gdf = gpd.GeoDataFrame(gdf, crs=sections.crs)

gdf.plot(alpha=0.3)
gdf.head()
```
